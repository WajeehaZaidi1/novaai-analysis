```python
def get_allowance(plan):
    if plan=='Basic':
        return 5000
    elif plan=='Pro':
        return 10000
    elif plan=='Enterprise':
        return 25000
totalover_by=0
for customer in customers:
    allowance = get_allowance(customer["plan"])
    if customer['ai_requests'] > allowance:
        over_by=customer['ai_requests'] - allowance
        totalover_by=over_by+1
        over_percentage = (over_by / allowance) * 100
        print(customer['name'], customer['plan'],allowance,over_by, over_percentage,'%')

    
    
```

```python
def calculate_total_revenue(customer_list):
    total = 0
    for customer in customers:
        total += customer["monthly_fee"]
    return total
def calculate_average_satisfaction(customer_list):
    total = 0
    for customer in customers:
        total += customer["satisfaction"]
    average = total / len(customer_list)
    return average
def get_risk_category(customer):

    if customer["satisfaction"] <= 5 and customer["tickets"] >= 8:
        return "HIGH RISK"
    elif customer["satisfaction"] <= 7 or customer["tickets"] >= 6:
        return "WATCH"
    else:
        return "HEALTHY"  
def get_ai_usage_percentage(customer):
    allowance = get_allowance(customer["plan"])
    percentage = (customer["ai_requests"] / allowance) * 100
    
    return percentage
def get_customers_over_allowance(customer_list):
    over_list = []
    for customer in customers:
        allowance = get_allowance(customer["plan"])
        if customer["ai_requests"] > allowance:
            over_list.append(customer)
    return over_list
def get_high_risk_customers(customer_list):
    high_risk_list = []
    for customer in customers:
        risk = get_risk_category(customer)
        if risk == "HIGH RISK":
            high_risk_list.append(customer)
    return high_risk_list
    
print(calculate_total_revenue(customers))
print(calculate_average_satisfaction(customers))
print(get_risk_category(customers[0]))
print(get_ai_usage_percentage(customers[0]))
print('no 5 ',get_customers_over_allowance(customers))    
print('no 6 ' ,get_high_risk_customers(customers))
```
